```python
#(1)掛け算のプログラム
for i in range(9):
    for j in range(9):
        print(f"{i+1}*{j+1}")
```

```python
#(2)奇数の段だけ計算
for i in range(9):
    if (i+1)%2==0:
        continue
    for j in range(9):
            print(f"{i+1}×{j+1}={(i+1)*(j+1)}")
```

```python
#(3)50以上になると次の段に進む
for i in range(9):
    if (i+1) % 2 == 0:
        continue
    for j in range(9):
        if (i+1)*(j+1) > 50:
            break
        print(f'{i+1}×{j+1}={(i+1)*(j+1)}')
```

```python
# (1)
temp = list()
for n in range(10):
    data = float(input(f'{n+1}個目のデータを入力 >>'))
    temp.append(data)

# (2)
for count in range(len(temp)):
    print(f'{count+8}時　{temp[count]}度')
# ※リストtempへのデータ登録を前提とする。

# (3)
temp_new = list()
for count in range(len(temp)):
    if count == 5:
        temp_new.append('N/A')
    else:
        temp_new.append(temp[count])
print(temp)
print(temp_new)
# ※リストtempへのデータ登録を前提とする。

# (4)
total = 0
for data in temp_new:
    if isinstance(data, float):
        total = total + data
print(total / (len(temp_new) - 1))
# ※リストtemp_newへのデータ登録を前提とする。
```

```python
# (1)
numbers = [1, 1]
data = sum(numbers)
count = 2
while data <= 1000:
    numbers.append(data)
    data = data + numbers[count-1]
    count += 1
print(numbers)

# (2)
ratios = list()
for count in range(len(numbers)):
    if count == len(numbers) - 1:
        break
    ratios.append(numbers[count+1] / numbers[count])
print(ratios)
# ※（1）の処理を前提とする。

# (3)
for count in range(len(ratios)):
    ratios[count] = int(ratios[count] * 1000) / 1000
print(ratios)
# ※（2）の処理を前提とする。
```
